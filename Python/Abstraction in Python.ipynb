### Abstraction in Python:
- Python  Abstraction is used to hide the implementation details and expose only necessary parts, making the code simpler and easier to interact with.
- For Eg - A Smart phone is a great real-life example of data abstraction you can make calls (or) take photos without knowing the how signals (or) storage works.Only essential features are work and hidden details are shown.

### Abstract Base Class:
- In Python, Abstract Base Class (ABC) is used to achieve data abstraction by defining a common interface for its subclasses.
- Abstract classes are created by using abc module and @abstractmethod decorator, allowing developers to enforce method implementation in subclasses while hiding complex internal logic.

```python
from abc import ABC, abstractmethod

class Greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass # Abstract method

class English(Greet):
    def say_hello(self):
        #return 'chaitanya'
        print('Hello! World')

c1 = English()
print(c1.say_hello())
```

### Components of Abstraction:
#### 1. Abstract Method:
- Abstract methods are method declarations without a body defined inside an abstract class.
- They act as placeholders that force subclasses to provide their own specific implementation, ensuring consistent structure across dervied classes.

```python
from abc import ABC, abstractmethod

class Person(ABC):
    @abstractmethod
    def person_data(self):
        pass # Abstract Method, no implementation here 
```

#### 3. Abstract Properties:
- Abstract properties work like abstract methods but are used for properties.
- These properties are declared by @property decorator and marked as abstract using @abstractmethod.
- Subclasses must be implemented by his properties.                                                           

```python
from abc import ABC, abstractmethod

class Student(ABC):
    @property
    @abstractmethod
    def student_data(self):
        pass # Abstract Method, must be implemented by subclasses.

class Student2(Student):
    @property
    def student_data(self):
        return 'chaitanya'

# Instantiate the concrete subclass
s1 = Student2()
print(s1.student_data)
```

#### 4. Abstract Class Instantiation:
- Abstract classes cannot be instantiated directly.
- This is because they contain one (or) more abstract methods (or) properties that lack implementations.
- Attempting to instantiate an abstract class results in a TypeError.
