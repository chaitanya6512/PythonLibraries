### Python Functions:
- A Function is a block of reusable code for designed to perform a specific task.
- Function improves readability, modularity, and code reusability.
### Python Function argument different types:
#### 1. Positional Arguments:
- Arguments are assigned based on their position.It is a common type of arguments.

```python
def Student(name,age):
    print(f"My name is {name}.I am {age} old.")
Student('Chaitanya',25)
```

#### 2. Default Argument:
- If the value is not provided, then the it uses default value.
- Used when a function has a default value for a parameter.

```python
def Student(name='Chaitanya',age=25):
    print(f"My name is {name}.I am {age} old.")
Student()
Student('Rohith',22)
```

#### 3. Keyword Arguments:
- Arguments are passing using parameter names.Order does not matter.
  

```python
def Student(name,age):
    print(f"My name is {name}.I am {age}.")
Student(name='Chaitanya', age=25)
Student(age=23,name='Rohith')
```

#### 4. Arbitary Arguments(args, *kwargs):
- args(Multiple Positional Arguments) - Allows passing multiple positional arguments as a tuple.


```python
def Sum_numbers(*num):
    total = sum(num)
    print(f"Sum:{total}")
Sum_numbers(1,2,3,4,5)
```

- *kwargs(Multiple Keyword Arguments) - Allows passing multiple keyword arguments as a dictionary.

```python
def Sum_numbers(**info):
    for key,value in info.items():
        print(f"{key}:{value}")
Sum_numbers(name="chaitanya",age=25,city='Chennai')
```

### Returning Values from Functions:
- A Function can return values using the return statement.

```python
def square(num):
    return num*num
result = square(5)
print(f"Square:{result}")
```

#### Returning Multiple values:


```python
def squ(a,b):
    return a+b,a-b,a*b,a/b
add, sub, mul, div = squ(2,3)
print(f"Addition:{add}")
print(f"Subtraction:{sub}")
print(f"Multiplication:{mul}")
print(f"Division:{div}")
```

#### Scope and lifetime of variables:
- local scope - Variables defined inside a function exist only within that function.
- global scope - Variables defined outside a function that can be accessed anywhere by using global keyword.

```python
a = 25 # global
def locals():
    a = 24 # local
    print(f"Inside function:",a)
locals()
print(f"Outside function:",a)
```

```python
a = 25
def globals():
    global a
    a = 56
    print(f"Inside function:",a)
globals()
print(f"Outer function:",a)
```

### Recursive Functions:
- A recursive function calls itself to solve a problem.
#### Factorial Using Recursion:

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)
result = factorial(5)
print(f" Factorial number : {result}")
```

#### Fibonacci Series using recursion:

```python
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
result = fibonacci(6)
print(f"Fibonacci Series:{result}")
```

### Anonymous Function(Lambda function):
- A lambda function is a small, anonymous function that can take any number of arguments but has only one expression.
- Syntax - lambda arguments : expression

```python
square = lambda x : x*x
print(square(5))
```

#### Multiple Arguments:


```python
sqa = lambda x, y :x + y
print(sqa(2,5))
```
