### Inheritance and Types in Python:
#### Inheritance:
- It is a key concept of OOPs that allows one class(child) to inherit the properties and methods of another class.
- This promotes code reusability and improves maintainability.
#### Types of Inheritance:
#### 1. Single Inheritance:
- It enables a dervied/ child class to inherit properties from a single parent class.

```python
class Parent:
    def func1(self):
        print("This func is in parent class.")
class Child(Parent):
    def func2(self):
        print("This function is in child class,")
obj = Child()
obj.func1()
obj.func2()
```

#### 2. Multiple Inheritance:
- It enables a derived/child class to inherit properties from multiple parent class.

```python
class Mother:
    mothername = ""
    def mother(self):
        print(self.mothername)
class Father:
    fathername = ""
    def father(self):
        print(self.fathername)
class Son(Mother, Father):
    def son(self):
        print(f"Father Name:{self.fathername}")
        print(f"Mother Name:{self.mothername}")
s1 = Son()
s1.fathername = 'Ramu'
s1.mothername = 'Sita'
s1.son()
```

#### 3. Multilevel Inheritance:
- It features of the base class and the dervied class are further inherited into new dervied class.

```python
class Grandfather:
    def grandfather(self):
        print('Iam grandfather')
class Parent(Grandfather):
    def parent(self):
        print('Iam parent')
class Child(Parent):
    def child(self):
        print('Iam child')
obj = Child()
obj.grandfather()
obj.parent()
obj.child()
    
```

```python
class Grandfather:
    def __init__(self, grandfathername):
        self.grandfathername = grandfathername
class Father(Grandfather):
    def __init__(self, grandfathername, fathername):
        self.fathername = fathername
        # Call the constructor
        Grandfather.__init__(self, grandfathername)
class Son(Father):
    def __init__(self, sonname, fathername, grandfathername):
        self.sonname = sonname
        Father.__init__(self, fathername, grandfathername)
    def print_names(self):
        print(f'Grandfather name:{self.grandfathername}')
        print(f'Father name:{self.fathername}')
        print(f'Son name:{self.sonname}')

s1 = Son('Sai', 'Chaitanya', 'Rohith')
print(s1.grandfathername)
s1.print_names()
```

#### 4. Hierarical inheritance:
- When morethan one derived class are created from a single base .

```python
class Parent:
    def func1(self):
        print("This func is in parent class.")
class Child1(Parent):
    def func2(self):
        print("This func is in Child1.")
class Child2(Parent):
    def func3(self):
        print("This func is in child2.")
obj1 = Child1()
obj2 = Child2()
obj1.func1()
obj1.func2()
obj2.func1()
obj2.func3()
```

### Use of super() Function:
- super - By using this function, you do not have to use the name of the parent element, it will automatically inherit the methods and properties from its parent.
- We will not specify self parameter inthe super function.

```python
# Without super() 
class Person:
  def __init__(self, fname, lname):
    self.firstname = fname
    self.lastname = lname

  def printname(self):
    print(self.firstname, self.lastname)

class Student(Person):
  def __init__(self, fname, lname):
    Person.__init__(self, fname, lname)

x = Student("Mike", "Olsen")
x.printname()
```

```python
#With super() function
class Person:
  def __init__(self, fname, lname):
    self.firstname = fname
    self.lastname = lname

  def print_name(self):
    print(self.firstname, self.lastname)

class Student(Person):
  def __init__(self, fname, lname):
    super().__init__(fname, lname)

x = Student("Mike", "Olsen")
x.print_name()
```
