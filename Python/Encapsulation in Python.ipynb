### Python Encapsulation:
- It is about protecting data inside a class.
- It means keeping data(properties) and methods together in a class, while controlling how the data can be accessed from outside class.
#### Private Properity:
- In python, you can make properties private by using a double underscore __prefix.
- To access private method, we can use getter method.
- To modify private method, we can use setter method.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.__age = age # Private Property
s1 = Student('chaitanya', 25)
print(s1.name)
print(s1.__age) # This will gives error
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age
    def get_age(self):
        return self.__age
p1 = Person('Chaitanya', 25)
print(p1.get_age())
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age
        
    def get_age(self):
        return self.__age
        
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print('Age must be positive')
            
p1 = Person('Chaitanya', 25)
print(p1.get_age())

p1.set_age(45)
print(p1.get_age())
```

#### Protect Properties:
- In python, You can make properties to protect by using single underscore _prefix.

```python
class Person:
  def __init__(self, name, salary):
    self.name = name
    self._salary = salary # Protected property

p1 = Person("Linus", 50000)
print(p1.name)
print(p1._salary) # Can access, but shouldn't
```
