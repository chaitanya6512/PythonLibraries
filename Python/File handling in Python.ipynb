### File handling in Python
- Python has several functions for creating, reading, updating and deleting files.
- The key function for working with files in python is the open() function.
- The open function takes two parameters;filname and mode.
- There different types of modes for opening a file.
  - 'r' - Read - Default value. Opens afile for reading, error if the file does not exist.
  - 'a' - Append - Opens a file for appending, creates a file if it does not exist.
  - 'w' - Write - Opens a file for writing, creates a file if it does not exist.
  - 'x' - create - Creates the specific  file, returns an error if the file exists.

```python
f = open("example.txt", "rt")
print(f.read())
```

#### Using "with" statement to open files and close files:
- It is a good pratice to close a file when you are done your work.


```python
with open("example.txt") as f:
    print(f.read())
f.close()
```

#### Read Only Parts of the File:
- read(5) - by specifying how many characters you want to return.

```python
with open('example.txt') as f:
    print(f.read(5))
```

- readline() - You can return one line by using the readline() method.

```python
with open('example.txt') as f:
    print(f.readline())
```

#### Python File Write
- 'w' - write - will overwrite any existing content.
- 'a' - Append - will append to the end of the file.

```python
with open('example.txt','a') as f:
    f.write('Now the file has more content')
with open('example.txt') as f:
    print(f.read())
```

#### Overwrite Existing Content
- use the 'w' parameter.

```python
with open("demofile.txt", "w") as f:
  f.write("Woops! I have deleted the content!")

#open and read the file after the overwriting:
with open("demofile.txt") as f:
  print(f.read())
f.close()
```

### Python Delete a File
- os.remove() - function To delete a file.

```python
import os
os.remove('example.txt')
```

#### Check if File exist:
- To avoid an error, you might want to check if the file exist are not.

```python
import os
if os.path.exists('example.txt'):
    os.remove('example.txt')
else:
    print('The file name not exist')
```

#### To Delete a Folder:
- os.rmdir('myfolder') - To delete entire folder.

```python
import os
os.rmdir('myfolder')
```
