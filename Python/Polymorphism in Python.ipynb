### Polymorphism in Python:
- In Python, Polymorphism allows same method, function, (or) operator to behave differently working on object it is working with.
- In other words, Polymorphism means same inferface(power button), but different behaviour based on device(object).
- This makes code more flexible and reusable.
- For Eg -  A remote control can operate multiple devices like TV, music, AC system.You press the power button and each device respond differently TV turns on, Ac starts cooling, music system plays music.

### Types of Polymorphism:
#### 1. Complie-time Polymorphism:
- It means deciding which method (or) operator run during compliation, usually through method (or) operator overloading.
- Language like c++ (or) Java support this.But Python doesn't because it's dynamically typed it resolves method calls at runtime, not during compilation.

```python
class Calculator:
    def multiply(self, a=1, b=1, *args):
        result = a * b
        for num in args:
            result *= num
        return result

cal = Calculator()
# using default arguments
print(cal.multiply())
print(cal.multiply(7))
# using multiple arguments
print(cal.multiply(2,3))
print(cal.multiply(2,3,4))
```

#### 2. Run-time Polymorphism:
- Run-time polymorphism means that the behavior of a method is decided while program is running, based on the object calling it.
- In Python, this happen through method overriding a child class provides its own version of a method already defined in the parent class.Since python is dynamic, it support this, allowing same method call to behave differently for different object types.

```python
class Animal:
    def sound(self):
        return "Some generic sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

animals = [Animal(), Dog(), Cat()]
for animal in animals:
    print(animal.sound())
```
