### Classes and Objects in Python:
- Classes and Objects are foundamental concepts in Python's object-oriented programming (OOP).
#### Classes in Python:
- A class is a blueprint (or) template used to create objects.It defines the attributes (data) and methods(functions) that the objects created from it will have.
  - Syntax - class class_name():
- For Eg - a Car class might define attributes are brand and mode.methods like start_engine().
#### Objects in Python:
- We can create any no of objects for class.
- An Object is a specific instance of a class.It holds actual data and can perform actions defined by the class.
  - Syntax
  -  objectname.method()
  -  objectname.variable
- For Eg - my_car = Car("BMW",'i7') creates an object my_car with specific values for 'brand' and 'model'.
#### Key differences:
- A class does not take memory until objects are created.
- Each objects stores its own data(intsance variables) and can call methods defined in the class.
- Changes to a class affect all future objects, but changing one object's data does not affect other's .

```python
#Create a class
class MyClass:
    x = 58
print(MyClass)
```

```python
#Create a object
p1 = MyClass()
print(p1.x)
```

#### Multiple Objects and Delete Objects:
- del - You can delete objects by using the del keyword.
- We can create multiple objects from the same class.

```python
class MyClass1:
    x = 25
p1 = MyClass1()
p2 = MyClass1()
p3 = MyClass1()
print(p1.x)
print(p2.x)
print(p3.x)

del p1
```

### Python __init__() Method:
- __init__() - This method is used for assign values to the object properties, (or) to perform operations.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
p1 = Student('Chaitanya', 24)
print(p1.name)
print(p1.age)
```

### Why Use __init__()?
- Without the __init__() Method, you would need to set properties manually for each object.

```python
class Person:
    pass
p1 = Person()
p1.name = 'Chaitanya'
p1.age = 25
print(p1.name)
print(p1.age)
```

```python
class Student:
    def __init__(self, name, age=22):
        self.name = name
        self.age = age
p1 = Student('Fazul')
p2 = Student('Chaitanya',26)
print(p1.name,p1.age)
print(p2.name, p2.age)
```

### Python self Parameter:
- self - a reference to the current instance of the class.
- It is used to access properties and methods that belongs to the class.
- self parameter must be the first parameter of any method on the class.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def greet(self):
        print("Hello, my name is "+self.name)
p1 = Student('Chaitanya', 25)
p1.greet()
```

### Why Use self?
- Without self, Python would not know which object's properties you want to access.

```python
class Student:
    def __init__(self, name, age=86):
        self.name = name
        self.age = age
    def Student1(self):
        print(self.name)
p1 = Student('Chaitanya',25)
p2 = Student('Rohith')

p1.Student1()
p2.Student1()
```

### self Does Not Have to Be Named "self"
- It does not have to be named self, you can call it whatever you like, but it has to be the first parameter of any method in the class.

```python
class Student:
    def __init__(abc, name, age):
        abc.name = name
        abc.age = age
    def greet(cdf):
        print(f"Hello, my name is :{cdf.name}.Iam {cdf.age}")
p1 = Student('chaitanya', 25)
p2 = Student('Rohith', 96)
p1.greet()
p2.greet()
```

### Accessing Properties with self:
- You can access any property of the class using self.

```python
class Car:
    def __init__(self, name, model):
        self.name = name
        self.model = model
    def display_info(self):
        print(f"{self.name} {self.model}")
c1 = Car("Toyato", 2021)
c2 = Car("BMW", 1978)
c1.display_info()
c2.display_info()
```

### Calling Methods with self:
- You can also call other methods within the class using self.

```python
class Person:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return "Hello, "+self.name
    def welcome(self):
        message = self.greet()
        print(message + "! Welcome to our website.")
p1 = Person("Tony")
p1.welcome()
```

### Python Class Properties:
#### Class Properties:
- Properties are variables that belong to a class.They store data for each object created from the class.
#### Access Properties:
- You can access properties using dot notation.
#### Modify Properties:
- You can modify the value of properties on objects.
#### Delete Properties:
- del - You can delete properties from objects using del keyword.

```python
class Person1:
    def __init__(self, name, age):
        self.name = name
        self.age = age
s1 = Person1('Chaitanya', 25)
print(f'My name is {s1.name}')
print(f'I am {s1.age} old')

s1.age = 74
print(f'I am {s1.age} old')

del s1.age
print(s1.age)
```

### Class Properties vs Object Properties
- Properties belongs inside ___init___() belong to each object(instance peoperties).
- Properties belongs outside methods belong to the class itself(class properties) and shared by all objects.


```python
class Person:
    species = 'Human'
    def __init__(self, name):
        self.name = name
p1 = Person('Chaitanya')
p2 = Person('Toyato')
print(p1.name)
print(p2.name)
print(p1.species)
print(p2.species)
        
```

### Modifying Class Properties and Add New Properties:
- When you modify a class property, it affects all objects.
- We can add new properties to existing objects.

```python
class Student:
    lastname = ""
    def __init__(self, name, age):
        self.name = name
        self.age = age
p1 = Student('chaitanya', 25)
p1.lastname = 'Yash'
print(p1.name)
print(p1.age)
print(p1.lastname)
p1.city = 'Nodia'
print(p1.city)
```

### Python Class Methods:
#### Class Methods:
- Methods are functions that belong to a class. They define the behavior of objects created from the class.
#### Methods with Parameters:
- Methods can accept parameters just like regular functions.



```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def Greed(self):
        print(f"My name is {self.name}. I am {self.age}")
s1 = Student('chaitanya', 25)
s1.Greed()

class Calculate:
    def add(self, a, b):
        return a+b
    def mul(self, a, b):
        return a * b
cal = Calculate()
print(cal.add(2,3))
print(cal.mul(2,3))

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def celebrate(self):
        self.age += 1
        print(f"My name is {self.name}")
p1 = Person('Chaitanya', 25)
p1.celebrate()
```

#### The__str__() Method:
- It is a special method that controls what is returned when the object is printed.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
p1 = Person('chaitanya', 78)
print(p1)

class Person1:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name} {self.age}"
p = Person1('Rohith', 23)
print(p)
```

#### Multiple Methods:

```python
class Playlists:
    def __init__(self, name):
        self.name = name
        self.songs = []
    def add_song(self, song):
        self.songs.append(song)
        print(f"ADDED:{song}")
    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)
            print(f"REMOVE:{song}")
    def show_songs(self):
        print(f"Playlist '{self.name}':")
        for song in self.songs:
            print(f" - {song}")
my_playlist = Playlists('Favourites')
my_playlist.add_song("Bohemian Rhapsody")
my_playlist.add_song("Stairway to Heaven")
my_playlist.show_songs()
```
