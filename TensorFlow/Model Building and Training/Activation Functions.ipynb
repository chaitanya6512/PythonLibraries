### Activation Function in TensorFlow?
- Activation Functions add non-linearity to deep learning models and allows them to learn complex patterns.
- TensorFlow's tf.keras.activations module provides a variety of activation functions to use in different scenarios.
- An Activation function is a mathematical transformation applied to the output of a neural network layer.
- It helps in -
  - Controlling the flow of information.
  - Enabiling networks to learn complex features.
### List of activation Functions in TensorFlow:
#### 1. ReLU(Rectified Linear Unit):
- ReLU activation functions outputs the input directly if it is positive; otherwise, it will output zero.
- This helps mitigate the vanishing gradient problem and improves training efficiency.
  - Function - tf.keras.activations.relu(x)

```python
import tensorflow as tf

# ReLU activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

#### 2. Sigmoid Function:
- Sigmoid function outputs values between 0 and 1, making it suitable for binary classification tasks.
- However, it is prone to the vanishing gradient problem, especially for deep networks.
  - Function - tf.keras.activations.sigmoid(x)

```python
import tensorflow as tf

# Sigmoid activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(25, activation='sigmoid', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

#### 3. Tanh(Hyperbolic Tangent):
- Tanh is smilar to Sigmoid but outputs values between -1 and 1.
- This allows for both positive and negative outputs, making it better for hidden layers than Sigmoid.
- However, it still suffers from the vanishing gradient problem.
  - Function - tf.keras.activations.tanh(x)

```python
import tensorflow as tf

# Tanh activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='tanh', input_shape=(25,)),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

#### 4. Softmax Function:
- Softmax is commonly used in the output layer of neural network for multi-class classification.
- It converts raw inputs into probabilities, with sum of probabilities equal to 1.
  - Function - tf.keras.activations.softmax(x)

```python
import tensorflow as tf

# Softmax activations
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation = 'relu', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

#### 5. Leaky ReLU:
- Leaky ReLU is a modified version of the ReLU function.
- It allows a small, non-zero output for negative input values, which can be help avoid the "dying ReLU" problem, where neurons never activate.
  - Function - tf.keras.activations.LeakyReLU(alpha=0.3)

```python
import tensorflow as tf

# LeakyReLU activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, input_shape=(32,)),
    tf.keras.layers.LeakyReLU(alpha=0.3),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accracy'])
```

#### 6. ELU(Exponential Linear Unit):
- ELU is similar to ReLU but has an exponential curve for negative values.
- It helps avoid the vanishing gradient problem and has better performance on deep neural networks.
  - Function - tf.keras.activations.elu(x, alpha=1.0)
                                                     

```python
import tensorflow as tf

# ELU activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(25, activation='elu', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

#### 7. Swish:
- Swish is a newer activation function proposed by Google, which has shown better performance than ReLU.
- Its variants in many cases. It is defined as f(x) = x. sigmoid(x).
  - Function - tf.keras.activations.swish(x)

```python
import tensorflow as tf

# Swish activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(46, activation='swish', input_shape=(25,)),
    tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

#### 8. Softplus:
- Softplus is a smooth approximation of the ReLU function.
- It's continous and differentiable, which helps avoid issues like dead neurons.
- It outputs values between 0 and positive infinity.
  - Function - tf.keras.activations.softplus(x)

```python
import tensorflow as tf

# Softplus activation
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation = 'softmax', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation = 'softmax')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```
