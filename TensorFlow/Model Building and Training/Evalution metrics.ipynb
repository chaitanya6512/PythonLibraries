### Evaluation metrics:
- It accesses the performance of machine learning models.
- In TensorFlow, these metrics help quantify how well the model is performing during training and after it has been trained.
#### 1. Accuracy:

#### 2. Precision:
- It is mainly used in classification tasks.
- It measure how many predicted positive labels were actually positive.
- It's particularly useful when the cost of false positives is high.
  - Function - tf.keras.metrics.Precision()
  

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation='sigmoid')
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', 
              metrics=[tf.keras.metrics.Precision()])

model.fit(x_train, y_train)
```

#### 3. Recall:
- It is mainly used in classification tasks, especially in situations where false negatives are more costly than false positives.
- It measures how many actual psotive labels were correctly identified by the model.
  - Function - tf.keras.metrics.Recall()

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(32,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', 
              metrics=[tf.keras.metrics.Recall()])

model.fit(x_train, y_train)
```

#### 4. F1 Score:
- F1 Score is the harmonic mean of precision and recall, providing a balanced evaluation metric for classification tasks.
- It is particularly useful when you need to balance both false positives and false negatives.

#### 5. Mean Squared Error(MSE):
- MSE is commonly used for regression tasks.
- It measures the average squared difference between the actual and predicted values.
- A lower MSE indicates better model performance.
  - Function - tf.keras.metrics.MeanSquaredError()

```python
import tensorflow as tf

model = tf.keras.Squential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(32,)),
    tf.keras.layers.Dense(10, activation='sigmoid')
])

model.compile(optimizer='adam', loss='mean_squared_error', 
              metrics=[tf.keras.metrics.MeanSquaredError()])

model.fit(x_train, y_train)
```
