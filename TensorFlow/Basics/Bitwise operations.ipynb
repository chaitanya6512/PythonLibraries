### What is Tensor Bitwise operations?
- Tensor bitwise operations in Python involve performing logical operations at the bit level on tensors, which are multi-dimensional arrays widely used inmachine learning and deep learning frameworks like TensorFlow and PyTorch.
### Tensor Bitwise Operations:
#### Bitwise AND Operation:
- tf.bitwise.bitwise_and() - Function is used in the code performs a bitwise AND operation element-wise between two TensorFlow tensors.

```python
import tensorflow as tf

# Define the tensors
tensor1 = tf.constant([5, 3, 8])
tensor2 = tf.constant([9, 8, 7])

result_and = tf.bitwise.bitwise_and(tensor1, tensor2)
print('Bitwise AND Operation:')
print(result_and)
```

#### Bitwise OR Operation:
- tf.bitwise.bitwise_or() - Function is used in the code performs a bitwise OR Operation element-wise between Two TensorFlow tensors.
- tf.bitwise.left_shift() - Function is applied to the result of the bitwise OR operation(result_or).

```python
import tensorflow as tf

# Define the Tensors
t1 = tf.constant([5,3,8])
t2 = tf.constant([7,1,9])
result_or = tf.bitwise.bitwise_or(t1,t2)

result_shift = tf.bitwise.left_shift(result_or, 2)

print('Result of Bitwise OR and shift Operation:')
print(result_shift.numpy())
```

#### Bitwise XOR with Bit Counting:
- tf.bitwise.bitwise_xor() - Function is used in the code performs a bitwise XOR Operation element-wise between two TensorFlow tensors.

```python
import tensorflow as tf

# Define the Tensors
t1 = tf.constant([5,7,3])
t2 = tf.constant([8,6,1])
result_xor = tf.bitwise.bitwise_xor(t1, t2)

count_bits = tf.math.count_nonzero(result_xor, axis=-1)
print('Result of Bitwise XOR and Counting set Bits:')
print(count_bits)
```

#### Bitwise Invert Operation:
- tf.bitwise.invert() - computes the complement of that element, swapping each 0 bit to 1 and each 1 bit to 0, regardless of the datatype of the tensor.

```python
import tensorflow as tf

# Create a sample tensor
tensor = tf.constant([1, 2, 3, 4, 5], dtype=tf.uint8)

# Invert the bits of the tensor
inverted_tensor = tf.bitwise.invert(tensor)

# Print the original and inverted tensors
print("Original tensor:", tensor.numpy())
print("Inverted tensor:", inverted_tensor.numpy())
```

#### Left and right shift operations:
- tf.bitwise.left_shift() - This operation shifts the bits of each element in the tensor to the left by a specified no of positions, padding the right side with zeros.
- tf.bitwise.right_shift() - This operation shifts the bits of each element in the tensor to the right by a specified no of positions, padding the right side with zeros.

```python
import tensorflow as tf

# Define the tensor
t1 = tf.constant([5,3,7])

result_left_shift = tf.bitwise.left_shift(t1,1)
print('Result of Left Shift Operation:')
print(result_left_shift.numpy())

result_right_shift = tf.bitwise.right_shift(t1,1)
print('Result of Right Shift Operation:')
print(result_right_shift.numpy())
```
