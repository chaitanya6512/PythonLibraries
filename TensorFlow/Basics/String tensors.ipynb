### What is string tensors?
- The string data type is used to represent string values. We use -
  - tf.string
- In this article, we will learn about the tf.string, exploring its operations, encoding/decoding techniques, comparison methods, real-world applications etc.
### How to create string tensors?


```python
import tensorflow as tf

# Create a sclaer string tensor
scaler_string_tensor = tf.constant("Hello, TensorFlow!")
print(scaler_string_tensor)

# Create a vector of tsring tensor
vector_string_tensor = tf.constant(["Hello", "TensorFlow", "World"])
print(vector_string_tensor)

# Create a 2D matrix of string tensor
matrix_string_tensor = tf.constant([["Hello", "World"], ["TensorFlow","Rocks!"]])
print(matrix_string_tensor)
```

### Operations can be performed by String tensors?
- It provides a set of string operations that can be used on tf.string tensors.
- It support many operations, including concatenation, splitting, padding, and indexing.
#### concatenation:
- We create two string constants using tensorflow, join them together with a space seperator, and then prints the result as numpy array.
  - tf.strings.join(t1,t2, separator=" ")

```python
import tensorflow as tf

t1 = tf.constant("Hello")
t2 = tf.constant("World!")
result = tf.strings.join([t1, t2],separator=" ")
print(result.numpy())
```

  - words = tf.strings.split(sentence)

#### Indexing:

  - sentence = tf.constant("Welcome to tensorflow")

  - char = tf.strings.unicode_split(sentence, "UTF-8")

  - print(char[0])

```python
import tensorflow as tf
sentence = tf.constant("Welcome to TensorFlow")
words = tf.strings.unicode_split(sentence, "UTF-8")
print(words)

import tensorflow as tf
sentence = tf.constant("Welcome to TensorFlow")
char = tf.strings.unicode_split(sentence, "UTF-8")
print('\n')
print(char[0])
```

### How strings tensor can be used for comparison and matching?
- string tensors can be compared for equality, similarity, or matched using the regular expresions with tf.strings functions like tf.strings.regmax_match.
#### Comparison:
- The code compares two strings str1 and str2 using TensorFlow's function to check if they are equal.
  - tf.strings.compare

```python
import tensorflow as tf
t1 = tf.constant(['Hello'])
t2 = tf.constant(['World'])
t3 = tf.equal(t1, t2)
print(t3)

t4 = tf.constant('Ten')
t5 = tf.constant('Ten')
t6 = tf.strings.regex_full_match(t4, t5)
print('\n')
print(t6)
```

### Working with Batched String Tensors:
- Efficiently handling batched string tensors is essential in many machine learning tasks.
#### Batching:
- The code splits a batch of sentences into words using -
  - tf.strings.split
#### Unbatching:
- The code joins the words in each sentence back into sentences using -
  - tf.strings.join

```python
import tensorflow as tf

# Define the Batching
batched_sentences = tf.constant(["TensorFlow is awesome", "Machine learning is fun"])
words = tf.strings.split(batched_sentences)
print(batched_sentences)
print('\n')
print(words)

# Define the Unbatching
unbatched_sentences = tf.strings.join(words[0], separator="")
print('\n')
print(unbatched_sentences.numpy())
```

### String Tensor Preprocessing in TensorFlow Models:
- Preprocessing string data is crucial before feeding it into TensorFlow models.
- Utilizes tf.strings functions like tf.strings.lower, tf.strings.regex_replace, etc.. for preprocessing tasks.
#### Convert to lower:
- We can convert to lower case by using the -
  - tf.strings.lower()
#### Convert to Upper:
- We can convert to upper case by using the -
  - tf.strings.upper()

```python
text = tf.constant("Hello, TensorFlow!")
processed_text1 = tf.strings.lower(text)
print(processed_text1)


processed_text2 = tf.strings.upper(text)
print(processed_text2)
```

### Handling missing values in string tensors:
- Strategies like using default values or special tokens are essential for handling missing or empty string values in Tensorflow.
- We can use handling empty or missing values will be -
  - tf.strings.replace

```python
str_with_missing = tf.constant("Hello Tensorflow <br /><b>contains string</b>")
str_with_default = tf.strings.regex_replace(str_with_missing, "<[^>]+>", "")
print(str_with_default)
```
